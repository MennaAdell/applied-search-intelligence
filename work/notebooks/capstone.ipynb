## 1. Question

*The research question and the decision it supports.*

# Decision-Support Search Intelligence: Balancing Content Refresh and Quick-Win Optimizations

**Abstract**: Search performance across large content portfolios suffers from decay masked by aggregate vanity metrics. We evaluate a tabular decision-support pipeline trained on 2026 production GCS search warehouse data (`FlyRank/internship-warehouse`). Using a leakage-free grouped split by client ID, random forest models achieve an honest generalization accuracy of 0.76 compared to a 0.65 rule-based baseline. Error analysis reveals trade-offs between long-tail noise and structural staleness. We deploy a human-in-the-loop action playbook prioritizing top-decile refresh and position-band tweaks. Results provide public-safe, decision-support guardrails for editorial teams.

---

* **Research Question**: Can we reliably prioritize search content interventions (refresh vs. quick-win metadata tweaks) using pre-decision historical telemetry without domain leakage?
* **Decision Supported**: Weekly editorial and SEO sprint prioritization queue.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

* **Assumptions**: Impression-to-click ratios locally proxy query-intent fit.
* **Features & Label**:
  * Features = historical impression/click volumes (`gsc_impressions`, `sessions_organic`).
  * Label = binary performance threshold (`gsc_impressions > 50`).
* **Baseline**: Rule-based threshold (`impressions < 50` $\rightarrow$ `REFRESH_CONTENT`).
* **Validation Design**: `GroupShuffleSplit` grouped by `client_id` to prevent cross-domain target leakage.
* **Leakage Audit**: Confirmed zero future-window columns or post-decision metrics in feature matrices.
