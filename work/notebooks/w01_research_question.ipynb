## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**The Question:** Which existing URLs have the highest statistical potential for a traffic lift if updated?  
**The Decision:** Prioritizing which specific pages the editorial team should focus on updating each week.  
**The Action:** Content creators will take the top 5 scored URLs from our model and actively refresh them.  
**The Cost of a Wrong Call:** Wasting limited human hours rewriting a page that yields zero traffic lift, while missing a genuinely decaying high-traffic page.  
**Why ML?** Simple sorting doesn't capture complex decay patterns or the relationship between high impressions and dropping click-through rates (CTR) over time.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Based on a quick look at our sample dataset, the numbers clearly validate this lane for the next weeks:
* **Total Content Opportunities:** 200 records analyzed in this sample.
* **Average Refresh Score:** 77.95, indicating a high overall urgency and potential traffic lift across these pages.
* **High Confidence Opportunities:** 161 out of 200 opportunities are flagged with 'high' confidence, proving that our model can reliably pinpoint exactly where the content team should focus their updating efforts.

```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/MennaAdell/applied-search-intelligence/main/outputs/refresh_queue_sample.csv')

total_opportunities = len(df)
print(f"Total Content Opportunities: {total_opportunities}")

avg_score = df['final_refresh_score'].mean()
print(f"Average Refresh Score: {avg_score:.2f}")

high_priority = df[df['confidence'] == 'high'].shape[0]
print(f"High Confidence Opportunities: {high_priority}")
```
