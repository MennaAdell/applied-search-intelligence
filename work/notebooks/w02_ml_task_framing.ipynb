## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Target / Proxy: We predict whether a page's performance is declining (is_declining_label == 1 based on trend_direction == "down"). This label comes from a defined operational rule over observed historical metrics (percentage change in traffic/impressions over time), serving as a reliable proxy for content refresh priority.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Success Metric: We use Precision@K (specifically Precision@50) to measure ranking quality, ensuring that the top-ranked pages recommended for content refresh are actually declining. A "good" number is achieving a Precision@50 score above 0.70, meaning at least 70% of the flagged pages genuinely require an update.
